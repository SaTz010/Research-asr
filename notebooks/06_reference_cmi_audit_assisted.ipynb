# Assisted reference and CMI review — 50 clips

Listen and correct the reference, generate script-based tag suggestions, then check and save. Suggestions are not verified language labels. No ASR predictions or API calls are used. Read `reports/reference_annotation_guide.md`. Items 1–25: November 11; 26–50: November 12. Existing reviews use the same annotation file; no review is saved automatically.


## 1. Setup and saved progress


```python
from pathlib import Path
import json, sys
from IPython.display import Audio, display
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src/reference_audit.py').is_file())
sys.path.insert(0, str(ROOT / 'src'))
from reference_audit import save_review
selection = json.loads((ROOT / 'data/manifests/reference_audit_20261105_selection.json').read_text(encoding='utf-8'))
progress_path = ROOT / 'data/annotations/reference_audit_20261105.json'
progress = json.loads(progress_path.read_text(encoding='utf-8'))
print('Reviewed:', sum(r['verdict'] != 'pending' for r in progress['reviews']), '/ 50')
from reference_audit import suggest_language_tags
from asr_common import normalize_text
drafts = {}
```

## 2. November 11: items 1–25


### Review 01 — audio, published reference and language tags
1. Play the audio. 2. Correct the quoted text in the draft cell and run it. 3. Check every suggested word/tag, add word-number overrides, set `tags_reviewed = True` and a verdict, then run the save cell. If you change the transcript, rerun the draft and recheck overrides.


```python
entry = selection['clips'][0]
identity = entry['identity']
print(identity['clip_id'])
print('Published reference:', identity['transcription'])
print('Published CMI:', identity['cmi'], '| channel:', identity['speaker_id'])
display(Audio(filename=str(ROOT / entry['audio_path'])))
print('Saved review:', json.loads(progress_path.read_text(encoding='utf-8'))['reviews'][0])
```

```python
# Edit the quoted text after listening. Keep quotes around the entire transcript.
corrected_reference = "त्यसो भए अब है त्यसो अब हामी के गर्ने विचरण हेर्ने त्यसका लागि हेर्नुहोस् तपाईँलाई नि जुन एउटा quantity को अरूसँग विचरण हेर्न सजिलो हुन्छ त्यहीहरू बनाएर सुरु गर्दा हुन्छ।"
drafts[1] = (corrected_reference, suggest_language_tags(corrected_reference))
for word_number, (word, tag) in enumerate(zip(normalize_text(corrected_reference).split(), drafts[1][1]), 1):
    print(f'{word_number:2d}  {word}  -> {tag}')
# Check Latin-script Nepali, transliterated English, names and mixed words carefully.
```

```python
reference, suggested = drafts[1]
token_languages = list(suggested)
tag_overrides = {}  # Only wrong tags: e.g. {19: 'en'} changes word 19 to English.
for word_number, tag in tag_overrides.items():
    if isinstance(word_number, bool) or not isinstance(word_number, int) or not 1 <= word_number <= len(token_languages):
        raise ValueError('Override word numbers must be in the displayed 1-based range.')
    token_languages[word_number - 1] = tag
tags_reviewed = True  # Change to True only after checking every word/tag.
verdict = 'match'  # match if unchanged, corrected if edited, unsure if unresolved.
notes = ''
if verdict != 'pending':
    if tags_reviewed is not True:
        raise ValueError('Check all tag suggestions and set tags_reviewed = True before saving.')
    print(save_review(progress_path, 1, reference, token_languages, verdict, notes, tagging_method='script_suggestions_human_checked'))
```
